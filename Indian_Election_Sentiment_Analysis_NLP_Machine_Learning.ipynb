```python
# Sentiment Analysis
# Now, I will get started with sentiment analysis. I will do it by using the Textblob package in Python. Here I will use this package to perform simple text classification in either positive or negative on the basis of sentiment analysis:

# 1 Polarity refers to the sentiment or emotion expressed in the text, ranging from negative to positive.

# 2 Subjectivity,suggests that the text contains some subjects or opinions, measures the extent to which the text is subjective or opinionated rather than objective
```

```python
textblob1 = TextBlob(modi['Tweet'][12])
textblob1.sentiment
textblob2 = TextBlob(rahul['Tweet'][200])
textblob2.sentiment
```

```python
def find_pol(review):
  return TextBlob(review).sentiment.polarity
```

```python
modi['Sentiment Polarity'] = modi['Tweet'].astype(str).apply(find_pol)
rahul['Sentiment Polarity'] = rahul['Tweet'].astype(str).apply(find_pol)
```

```python
modi.head(3)
```

```python
rahul.head(3)
```

```python
# Sentiment Polarity on Both the candidates:
# Now I will add a new attribute in both the datasets by the name of “Expression Label”:
# 1 polarity > 0 :: positive
# 2 polarity < 0 :: negative
# 3 Polarity == 0 :: neutral
```

```python
modi['Expression label'] = np.where(modi['Sentiment Polarity']>0, 'positive', 'negative')
modi['Expression label'][modi['Sentiment Polarity']==0]=="Neutral"
print(modi.tail())

rahul['Expression label'] = np.where(rahul['Sentiment Polarity']>0, 'positive', 'negative')
rahul['Expression label'][rahul['Sentiment Polarity']==0]=="Neutral"
print(rahul.tail())
```

```python
neutral_modi = modi[modi['Sentiment Polarity']==0.0000]

remove_neutral_modi = modi['Sentiment Polarity'].isin(neutral_modi['Sentiment Polarity'])
modi.drop(modi[remove_neutral_modi].index, inplace=True)
```

```python
print(neutral_modi.shape)
print(modi.shape)
```

```python
neutral_rahul = rahul[rahul['Sentiment Polarity']==0.0000]

remove_neutral_rahul = rahul['Sentiment Polarity'].isin(neutral_rahul['Sentiment Polarity'])
rahul.drop(rahul[remove_neutral_rahul].index, inplace=True)
```

```python
print(neutral_rahul.shape)
print(rahul.shape)
```

```python
# balance both the datasets
#modi
np.random.seed(10)
remove_n = 8482
drop_indices = np.random.choice(modi.index, remove_n, replace=False)
df_subset_modi = modi.drop(drop_indices)
print(df_subset_modi.shape)
```

```python
np.random.seed(10)
remove_n = 367
drop_indices = np.random.choice(rahul.index, remove_n, replace=False)
df_subset_rahul = rahul.drop(drop_indices)
print(df_subset_rahul.shape)
```

```python
#Both modi and rahul are equal data is balace now.
```

```python
#Predict the India Elections, by analyzing the number of positive and negative sentiments in both the accounts:
```

```python
count_1 = df_subset_modi.groupby('Expression label').count()
print(count_1)
negative_per1 = (count_1['Sentiment Polarity'][0]/1000)*100
positive_per1 = (count_1['Sentiment Polarity'][1]/1000)*100

count_2 = df_subset_rahul.groupby('Expression label').count()
print(count_2)
negative_per2 = (count_2['Sentiment Polarity'][0]/1000)*100
positive_per2 = (count_2['Sentiment Polarity'][1]/1000)*100
```

```python
#plotting column graph for visual analysis


Politicians = ['modi', 'rahul']
lis_pos = [positive_per1, positive_per2]
lis_neg = [negative_per1, negative_per2]

fig = go.Figure(data=[
    go.Bar(name='Positive', x=Politicians, y=lis_pos),
    go.Bar(name='Negative', x=Politicians, y=lis_neg)
])
# Change the bar mode
fig.update_layout(barmode='group')
fig.show()
```
